```python
import csv
with open(titanic_path(), encoding='utf-8') as f:
    reader = csv.DictReader(f)
    rows = list(reader)
print(len(rows))
print(rows[0]['Name'], rows[0]['Age'], rows[0]['Fare'])
# 두번째 예제
total_fare = 0
for row in rows:
    total_fare += float(row['Fare'])
print(total_fare)
```

```python
first_class = []
for row in rows:
    if row['Pclass'] == '1':
        first_class.append({'Name': row['Name'], 'Fare': row['Fare']})
out = output_path('first_class.csv')
with open(out, 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=['Name', 'Fare'])
    writer.writeheader()
    writer.writerows(first_class)
```

```python
# 타이타닉 데이터에서 Survived (1 = 생존, 0 = 죽음)
# 총 몇명 살았니? 
survived = 0
for row in rows:
    if row['Survived'] == '1':
        survived += 1
print(survived)
# ----
# Age == '' <- 이거 몇건이니? 
missing = 0 
for row in rows:
    if row['Age'] == '':
        missing += 1
print(missing)
```
