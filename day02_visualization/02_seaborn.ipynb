```python
import seaborn as sns
import matplotlib.pyplot as plt
from helper import load_titanic, titanic_corr

df = load_titanic()
```

```python
sns.countplot(data=df, x='Pclass', hue='Sex')
plt.title('Passengers by Class and Sex')
```

```python
sns.barplot(data=df, x='Pclass', y='Survived', hue='Sex', errorbar=None)
plt.title('Survival Rate by Class and Sex')
```

```python
sns.boxplot(data=df, x='Pclass', y='Age')
```

# 3:10 까지 휴식

```python
sns.histplot(data=df, x='Age', bins=20, kde=True)
plt.title('Age Distribution')
```

```python
sns.kdeplot(data=df, x='Age', hue='Survived', fill=True)
plt.title('Age by Survival')
```

```python
corr = titanic_corr()
plt.figure(figsize=(6, 5))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)
```

```python
small = df[['Age', 'Fare', 'Pclass', 'Survived']].sample(200, random_state=0)
sns.pairplot(small, vars=['Age', 'Fare', 'Pclass'], hue='Survived', height=2.2)
```
