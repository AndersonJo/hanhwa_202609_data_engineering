# Z-Score: 평균에서 표준편차가 몇 배 떨어졌나

```python
df['z'] = (df['Fare'] - df['Fare'].mean()) / df['Fare'].std()
# df[df['z'] >= 2.5][['Fare', 'z']]
```

```python
q1 = df.Fare.quantile(0.25)
q3 = df.Fare.quantile(0.75)
iqr = q3 - q1
print('q1:', q1, 'q3:', q3, 'iqr:', iqr)
low = q1 - 1.5 * iqr
high = q3 + 1.5 * iqr

is_outlier = (df.Fare < low) | (df.Fare > high)
df[is_outlier].Fare
```
