```python
from helper import load_titanic

df = load_titanic()
df.sort_values('Fare', ascending=False)[['Name', 'Fare']].head(3)
```

```python
df.sort_values(['Pclass', 'Age'], ascending=[True, False])[['Name', 'Pclass', 'Age']]
```

```python
df.groupby('Sex')['Survived'].mean().round(2)
```

```python
df.groupby('Pclass')['Fare'].mean()
```

```python
df['ClassAvgFare'] = df.groupby('Pclass')['Fare'].transform('mean')
# 여러분이 할것: FareRatio => Fare / ClassAvgFare
df['FareRatio'] = df['Fare'] / df['ClassAvgFare']
df[['Name', 'Pclass', 'Fare', 'ClassAvgFare', 'FareRatio']]
```
