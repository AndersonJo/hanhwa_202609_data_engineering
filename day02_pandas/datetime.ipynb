```python
import pandas as pd
from helper import load_bike_raw

df = load_bike_raw()
df['date'].dtype
df['date'] = pd.to_datetime(df['date'], format='%Y-%m-%d')
print(df['date'].dtype)
```

```python
df['date'].dt.year.head(2).tolist()
df['date'].dt.month.head(2).tolist()
df['date'].dt.dayofweek.head(2).tolist()
df['date'].dt.day_name().head(2).tolist()
```

```python
df['datetime'] = df['date'] + pd.to_timedelta(df['hour'], unit='h')
df[['date', 'hour', 'datetime', 'rented']]
```

```python
df.groupby(df['date'].dt.month)['rented'].sum()
df.groupby('hour')['rented'].mean().sort_values(ascending=False)
# hour <- 시간대별 평균 rented 수 계산하는데.. -> 정렬  내리차순
```

```python
df['is_weekend'] = df['date'].dt.dayofweek >= 5
df.groupby('is_weekend')['rented'].mean()
```

- nlargest 함수 <- 구글 검색

```python
season_mask = df['season'].isin(['Spring', 'Autumn'])
temperature_mask = df['temp'].between(20, 25)

df[season_mask & temperature_mask].groupby('hour')[['rented']].sum()
print()
```
